#Q17. Sales Data Analysis

```python
df = pd.read_csv("/content/drive/MyDrive/DSAB13/Assessment/sales.csv")
df
```

```python
import pandas as pd
import numpy as np

sales_df = pd.read_csv("/content/drive/MyDrive/DSAB13/Assessment/sales.csv")

# Clean Sales column: remove commas and convert to numeric
sales_df['Sales'] = sales_df['Sales'].astype(str).str.replace(',', '', regex=True)
sales_df['Sales'] = pd.to_numeric(sales_df['Sales'], errors='coerce')

# Clean Discount column: remove percentage signs or non-numeric characters and convert to numeric
sales_df['Discount'] = sales_df['Discount'].astype(str).str.replace('%', '', regex=True)
sales_df['Discount'] = pd.to_numeric(sales_df['Discount'], errors='coerce')

# Convert Order_Date into appropriate Pandas datetime type
sales_df['Order_Date'] = pd.to_datetime(sales_df['Order_Date'], format='mixed')

# Check for missing values and handle them
numerical_cols = sales_df.select_dtypes(include=[np.number]).columns
for col in numerical_cols:
    if sales_df[col].isnull().any():
        sales_df[col] = sales_df[col].fillna(sales_df[col].mean())

# B. Sales Analysis
total_sales = sales_df['Sales'].sum()
total_profit = sales_df['Profit'].sum()
avg_sales = sales_df['Sales'].mean()
top_5_products = sales_df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(5)
region_sales = sales_df.groupby('Region')['Sales'].sum()
highest_sales_region = region_sales.idxmax()

print("=== B. SALES ANALYSIS ===")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Average Sales: ${avg_sales:,.2f}")
print("\nTop 5 Products based on Sales:")
print(top_5_products)
print(f"\nRegion generating the highest sales: {highest_sales_region} (${region_sales.max():,.2f})")

# C. Category Analysis
category_summary = sales_df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Discount=('Discount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

print("\n=== C. CATEGORY ANALYSIS ===")
display(category_summary)

# D. Business Insights
most_profitable_category = category_summary['Total_Profit'].idxmax()
top_product = sales_df.groupby('Product')['Sales'].sum().idxmax()

print("\n=== D. BUSINESS INSIGHTS ===")
print(f"1. Most profitable category: {most_profitable_category}")
print(f"2. Region with the highest sales: {highest_sales_region}")
print(f"3. Product with the highest sales: {top_product}")
print("\nObservations:")
print("  - Computers drive the most value, contributing the highest share of both overall sales and profit.")
print("  - The North region is our key geographic market, bringing in the highest sales volume.")
print("  - Profitability remains robust even when average discount rates vary across product categories.")
```
