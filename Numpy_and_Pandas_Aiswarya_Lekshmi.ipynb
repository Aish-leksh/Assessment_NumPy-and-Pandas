#Q16. Student Performance Analysis

Use student_performance.csv .
A. Data Cleaning – 2 Marks
Identify missing values.
Handle missing numerical values appropriately.


```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/drive/MyDrive/DSAB13/Assessment/student_performance.csv")

print("Missing values:")
print(df.isnull().sum())

numerical_cols = df.select_dtypes(include=[np.number]).columns
for col in numerical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mean())

print("\nMissing values per column after filling numerical missing values:")
print(df.isnull().sum())

df

df['Average_Score'] = (df['Python_Score']+
                       df['SQL_Score']+
                       df['Pandas_Score']+
                       df['NumPy_Score'])/4

df['Pass_Status'] = np.where(df['Average_Score'] >= 50, 'Pass', 'Fail')


print(df)

print(df.sort_values(by='Average_Score', ascending=False).head(5))

dept_avg = df.groupby('Department')['Average_Score'].mean()
print(dept_avg.idxmax(), dept_avg.max())

print(df.groupby('Department')['Attendance'].mean())

print(df['Pass_Status'].value_counts())

summary = df.groupby('Department').agg(
    Total_Students=('Department', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
).reset_index()

print(summary)
```

#Q18. Advanced NumPy and Pandas Analysis

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/DSAB13/Assessment/employee.csv')

df['Salary'] = df['Salary'].astype(str).str.replace(',', '', regex=True)
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')

df['Experience'] = pd.to_numeric(df['Experience'], errors='coerce')


df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].mean())
df['Experience'] = df['Experience'].fillna(df['Experience'].mean())

salary = df['Salary'].to_numpy()
performance = df['Performance_Score'].to_numpy()

print("Mean Salary:", np.mean(salary))
print("Median Salary:", np.median(salary))
print("Std Dev of Salary:", np.std(salary))
print("Mean Performance Score:", np.mean(performance))

filtered = df[(df['Experience'] > 3) & (df['Performance_Score'] >= 80)]
print("\nFiltered Employees (Experience > 3 and Performance >= 80):")
print(filtered)

dept = df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Average_Performance_Score=('Performance_Score', 'mean'),
    Total_Projects_Completed=('Projects_Completed', 'sum'),
    Number_of_Employees=('Employee_ID', 'count')
)
print("\nDepartment Aggregations:")
print(dept)

print("\nTop Department by Performance Score:", dept['Average_Performance_Score'].idxmax())

result = df[(df['Salary'] > np.mean(salary)) &
            (df['Performance_Score'] > np.mean(performance))]
print("\nEmployees with above-average Salary and Performance:")
print(result[['Employee_ID', 'Department', 'Salary', 'Performance_Score']])
```
